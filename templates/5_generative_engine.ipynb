## Setup

```python
import getpass
import os

from moth_atlas import Atlas

if not os.environ.get("MOTH_API_KEY"):
    os.environ["MOTH_API_KEY"] = getpass.getpass("Atlas API key")

client = Atlas()  # reads MOTH_API_KEY
```

## Try it

Train once on an example, then generate from the trained model as often as you like.

```python
job = client.jobs.run(
    engine="TODO-train-v1",
    params={
        "TODO": "TODO example".split(),  # the example to learn from
        "TODO_param": 10,
    },
)
model = job.download(slot="TODO_model", to="model.json")[0]  # the slot that carries the model into generation
```

```python
job = client.jobs.run(
    engine="TODO-gen-v1",
    params={
        "TODO_param": 0.5,
    },
    files={
        "TODO_model": model,
    },
)
job.show()  # TODO: show the output the way a reader understands it
```
