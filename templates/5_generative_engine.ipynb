## Setup

```python
import getpass
import os

from moth_atlas import Atlas

if not os.environ.get("MOTH_API_KEY"):
    os.environ["MOTH_API_KEY"] = getpass.getpass("Atlas API key")

client = Atlas()  # reads MOTH_API_KEY
TRAIN_ENGINE = "TODO-train-v1"
GENERATE_ENGINE = "TODO-gen-v1"
MODEL_SLOT = "TODO"  # the slot that carries the model, out of training and into generation
```

## Try it

Train once on an example, then generate from the trained model as often as you like.

```python
train_params = {"TODO": "TODO example".split(), "TODO_param": 10}  # the example and the training parameters

job = client.jobs.submit(TRAIN_ENGINE, params=train_params)
client.jobs.wait(job.job_id, on_update=lambda s: print(s.status))
model = client.jobs.download(job.job_id, slot=MODEL_SLOT, to="model.json")
model
```

```python
params = {"TODO_param": 0.5}  # the generation parameters

model_asset = client.assets.upload(model)
job = client.jobs.submit(GENERATE_ENGINE, params=params, input_files={MODEL_SLOT: model_asset.asset_id})
client.jobs.wait(job.job_id, on_update=lambda s: print(s.status))
client.jobs.download(job.job_id)  # TODO: show where the engine puts its output
```
