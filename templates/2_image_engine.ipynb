## Setup

```python
import getpass
import os

from moth_atlas import Atlas

if not os.environ.get("MOTH_API_KEY"):
    os.environ["MOTH_API_KEY"] = getpass.getpass("Atlas API key")

client = Atlas()  # reads MOTH_API_KEY
ENGINE = "TODO-v1"
```

```python
from IPython.display import Image, display


def show(images, width=200):
    """Display {label: image bytes}."""
    for label, data in images.items():
        print(label)
        display(Image(data, width=width))
```

## Try it

Point `IMAGE` at your own image, or keep the sample beside this notebook.

```python
IMAGE = "TODO_sample.png"  # TODO: put a small sample beside the notebook

image = open(IMAGE, "rb").read()
show({"input": image})
```

```python
params = {"TODO_param": 0.5}  # the engine's parameter names and values

asset = client.assets.upload(IMAGE)
job = client.jobs.submit(ENGINE, params=params, input_files={"TODO_slot": asset.asset_id})
client.jobs.wait(job.job_id, on_update=lambda s: print(s.status))
show({"input": image, "output": client.jobs.download(job.job_id)})
```
