## Setup

```python
import getpass
import os

from moth_atlas import Atlas

if not os.environ.get("MOTH_API_KEY"):
    os.environ["MOTH_API_KEY"] = getpass.getpass("Atlas API key")

client = Atlas()  # reads MOTH_API_KEY
```

```python
import base64
from IPython.display import HTML, display

def row(images, width=170):
    """Show {label: image bytes} side by side."""
    display(HTML("".join(
        f'<figure style="display:inline-block;margin:4px;text-align:center">'
        f'<img src="data:image/png;base64,{base64.b64encode(data).decode()}" width="{width}"><figcaption>{label}</figcaption></figure>'
        for label, data in images.items())))
```
